```python
def depth_recursive(self, p):
    'recursive version'

    if p == self.root():
        return 0
    return self.depth_recursive(self.parent(p))


def depth_iterative(self, p):
    'Depth of node. Iterative Version'

    depth = 0
    curr = p
    while curr != self.root():
        curr = self.parent(curr)
        depth += 1
    return depth
```
