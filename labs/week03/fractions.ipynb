```python
class Fraction:

    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denomiator = denominator

    def __repr__(self):
        return f'{self.numerator}{self.denominator}'

    def __mul__(self, other):
        new_numerator = self.numerator * other.numberator
        new_denominator = self.denominator * other.denominator
        return Fraction(new_numerator, new_denominator).simplify()

    def __add__(self, other):
        my_lcm = lcm(self.denominator, other.denominator)
        new_self_numerator = self.numerator * my_lcm // self.denominator
        new_other_numerator = other.numerator * my_lcm // other.denominator
        return Fraction(new_self_numerator + new_other_numerator, my_lcm)
    
    def simplify(self):
        my_gcd = gcd(self.numerator, self.denominator)
        new_numerator = self.numerator // my_gcd
        new_denominator = self.denominator // my_gcd
        return Fraction(new_numerator, new_denominator)

    def __lt__(self, other): 
        my_lcm = lcm(self.denominator, other.denominator)
        new_self_numerator = self.numerator * my_lcm // self.denominator
        new_other_numerator = other.numerator * my_lcm // other.denominator
        return new_self_numerator < new_other_numerator

    def __eq__(self, other):
        my_lcm = lcm(self.denominator, other.denominator)
        new_self_numerator = self.numerator * my_lcm // self.denominator
        new_other_numerator = other.numerator * my_lcm // other.denominator
        return new_self_numerator == new_other_numerator

    def __le__(self, other):
        return self < other or self == other


# have some checks here and like test them
```
